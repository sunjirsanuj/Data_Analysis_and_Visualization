2. Create two NumPy arrays of shape (2, 3) and (3, 2). Perform matrix multiplication on these arrays.

```python
import numpy as np
```

```python
arr1 = np.random.randint(1, 31, size=(2, 3))
arr2 = np.random.randint(1, 31, size=(3, 2))
print(arr1, arr2)
```

```python
result = np.matmul(arr1, arr2)
print(result)
```
