1. Create a NumPy array of shape (5, 5) filled with random integers. Compute the mean, median, standard deviation, and variance of the array.

```python
import numpy as np
```

```python
arr = np.random.randint(1, 31, size=(5, 5))
print(arr)
```

```python
mean = np.mean(arr)
median = np.median(arr)
std = np.std(arr)
var = np.var(arr)

print(mean)
print(median)
print(std)
print(var)
```
