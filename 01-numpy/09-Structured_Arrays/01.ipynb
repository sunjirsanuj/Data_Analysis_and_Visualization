1. Create a structured array with fields 'name' (string), 'age' (integer), and 'weight' (float). Add some data and sort the array by age.

```python
import numpy as np
```

```python
people_dtype = [("name", "U20"), ("age", "i4"), ("weight", "f8")]

people = np.array([
    ("Sunjir", 21, 69.0),
    ("Karim", 24, 70.2),
    ("Azim", 21, 65),
    ("Arafat", 22, 50)
], dtype=people_dtype)
print(people)
```

```python
indices = np.argsort(people["age"])
sorted_byAge = people[indices]
print(sorted_byAge)
```
