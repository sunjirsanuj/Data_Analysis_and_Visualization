2. Create a NumPy array of shape (5, 5) filled with random integers. Flatten the array and then reshape it back to (5, 5).

```python
import numpy as np
```

```python
arr = np.random.randint(1, 31, size=(5, 5))
print(arr)
```

```python
flat_arr = arr.flatten()
print(flat_arr)
reshape_arr = flat_arr.reshape(5, 5)
print(reshape_arr)
```
