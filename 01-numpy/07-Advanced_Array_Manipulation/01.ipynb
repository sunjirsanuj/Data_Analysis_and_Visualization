1. Create a NumPy array of shape (3, 3) with values from 1 to 9. Reshape the array to shape (1, 9) and then to shape (9, 1).

```python
import numpy as np
```

```python
arr = np.arange(1, 10).reshape(3, 3)
print(arr)
```

```python
arr_1_9 = arr.reshape(1, 9)
arr_9_1 = arr.reshape(9, 1)
print(arr_1_9)
print(arr_9_1)
```
