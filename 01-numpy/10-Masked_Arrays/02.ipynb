2. Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements.

```python
import numpy as np
```

```python
arr = np.random.randint(1, 31, size=(3, 3))
diagonal_mask = np.eye(3, dtype=bool)
```

```python
print(arr)
print(diagonal_mask)
```

```python
masked_arr = np.ma.masked_array(arr, mask=diagonal_mask)
print(masked_arr)
```

```python
mean = masked_arr.mean()
print(mean)
filled = masked_arr.filled(mean)
print(filled)
```
