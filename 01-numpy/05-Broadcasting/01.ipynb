1. Create a NumPy array of shape (3, 3) filled with random integers. Add a 1D array of shape (3,) to each row of the 2D array using broadcasting.

```python
import numpy as np
```

```python
arr = np.random.randint(1, 31, size= (3, 3))
arr_1d = np.array([10, 20, 30])
print(arr)
print(arr_1d)
```

```python
result = arr + arr_1d
print(result)
```
