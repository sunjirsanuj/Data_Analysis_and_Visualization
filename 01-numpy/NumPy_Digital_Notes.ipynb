## Assignment 1 — Creating and changing array elements

### A. Replace an entire column
NumPy uses zero-based indexing: column index `0` is the first column, `1` the second, and `2` the third. `array[:, 2]` means **all rows, column index 2**. Assigning `1` changes every element selected.

```python
arr = np.random.randint(1, 21, size=(5, 5))
print("Before:")
print(arr)

arr[:, 2] = 1
print("After replacing the third column:")
print(arr)
```

### B. `np.arange()` and `reshape()`

- `np.arange(start, stop)` generates values from `start` up to (but not including) `stop`.
- `.reshape(rows, columns)` changes the array's shape without changing the order of its values. The new shape must contain the same number of elements.

For values 1 through 16, use `np.arange(1, 17)`, then reshape to `(4, 4)`.

```python
arr = np.arange(1, 17).reshape(4, 4)
print("Before:")
print(arr)

np.fill_diagonal(arr, 0)
print("After setting the diagonal to zero:")
print(arr)
```

### C. `np.fill_diagonal()`

`np.fill_diagonal(array, value)` replaces the main diagonal values (top-left to bottom-right). It modifies the given array directly.

## Assignment 2 — Indexing and slicing

### A. Extract a sub-array

General slicing format: `array[row_start:row_stop, col_start:col_stop]`.

The start index is included, but the stop index is excluded. Therefore, rows 3–5 and columns 2–4 (human counting) correspond to indices `2:5` and `1:4`.

```python
arr = np.arange(1, 37).reshape(6, 6)
print(arr)

sub_array = arr[2:5, 1:4]
print("Selected sub-array:")
print(sub_array)
```

### B. Extract border elements with slicing and concatenation

- `arr[0, :]`: first row.
- `arr[-1, :]`: last row.
- `arr[1:-1, 0]`: left edge excluding the corners.
- `arr[1:-1, -1]`: right edge excluding the corners.
- `np.concatenate((a, b, ...))`: joins arrays along an existing axis.

This approach returns the border in a clockwise-style sequence without repeating corners.

```python
arr = np.random.randint(1, 10, size=(5, 5))
print("Array:")
print(arr)

border = np.concatenate((
    arr[0, :],       # top edge, including both corners
    arr[1:-1, -1],   # right edge, excluding corners
    arr[-1, ::-1],   # bottom edge, right to left
    arr[-2:0:-1, 0]  # left edge, bottom to top, excluding corners
))
print("Border elements:")
print(border)
```

## Assignment 3 — Element-wise operations and sums

### A. Element-wise arithmetic

When two arrays have the same shape, `+`, `-`, `*`, and `/` operate on matching positions. This is element-wise arithmetic, not matrix multiplication.

- `+`: addition
- `-`: subtraction
- `*`: multiplication
- `/`: division (usually produces floating-point values)

Using `np.random.randint(1, 10, ...)` keeps values nonzero, so division by zero is avoided in this example.

```python
a = np.random.randint(1, 10, size=(3, 4))
b = np.random.randint(1, 10, size=(3, 4))

print("A:\n", a)
print("B:\n", b)
print("Addition:\n", a + b)
print("Subtraction:\n", a - b)
print("Multiplication:\n", a * b)
print("Division:\n", a / b)
```

### B. `np.sum()` and the `axis` argument

`np.sum(array)` adds every element.

For a 2D array:
- `np.sum(array, axis=1)` sums across columns, giving one total per **row**.
- `np.sum(array, axis=0)` sums down rows, giving one total per **column**.

A helpful rule: the axis being summed is removed from the output shape.

```python
arr = np.arange(1, 17).reshape(4, 4)
print(arr)
print("Row-wise sums:", np.sum(arr, axis=1))
print("Column-wise sums:", np.sum(arr, axis=0))
print("Total sum:", np.sum(arr))
```

## Assignment 4 — Statistical operations

### A. Mean, median, standard deviation, and variance

- `np.mean(array)`: arithmetic average.
- `np.median(array)`: middle value after sorting (or average of the two middle values).
- `np.std(array)`: standard deviation, describing spread around the mean.
- `np.var(array)`: variance, the average squared deviation from the mean under NumPy's default convention.

By default, `std` and `var` use `ddof=0` (population-style calculation). For sample statistics, use `ddof=1`.

```python
arr = np.random.randint(1, 101, size=(5, 5))
print(arr)
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard deviation:", np.std(arr))
print("Variance:", np.var(arr))
```

### B. Normalize values to mean 0 and standard deviation 1

Standardization uses:

`normalized = (array - mean) / standard_deviation`

Subtracting the mean centers values around zero. Dividing by the standard deviation scales their spread. For a constant array, standard deviation is zero, so this formula would divide by zero; the example below has varying values.

```python
arr = np.arange(1, 10).reshape(3, 3)
mean = np.mean(arr)
std = np.std(arr)
normalized = (arr - mean) / std

print("Original:\n", arr)
print("Normalized:\n", normalized)
print("New mean:", np.mean(normalized))
print("New standard deviation:", np.std(normalized))
```

## Assignment 5 — Broadcasting

**Broadcasting** lets NumPy perform operations between compatible shapes without manually copying the smaller array.

### A. Add a `(3,)` array to every row of a `(3, 3)` array

A 1D array of shape `(3,)` aligns with the last dimension (the columns), so `arr + row_values` adds those three values to each row.

```python
arr = np.random.randint(1, 10, size=(3, 3))
row_values = np.array([10, 20, 30])

print("Array:\n", arr)
print("Values added to each row:", row_values)
print("Result:\n", arr + row_values)
```

### B. Subtract a `(4,)` array down the rows / from each row across all columns

To subtract `1`, `2`, `3`, and `4` from the corresponding rows, convert the 1D array into a column-shaped array.

- `np.newaxis` inserts a new dimension.
- `values[:, np.newaxis]` changes shape `(4,)` into `(4, 1)`.
- The `(4, 1)` array broadcasts across the four columns.

This means the first value is subtracted from every item in row 1, the second from every item in row 2, and so on.

```python
arr = np.random.randint(1, 10, size=(4, 4))
values = np.array([1, 2, 3, 4])

print("Original shape:", values.shape)
column_values = values[:, np.newaxis]
print("After newaxis:", column_values.shape)
print("Column-shaped values:\n", column_values)
print("Original array:\n", arr)
print("Result:\n", arr - column_values)
```

## Assignment 6 — Linear algebra

### A. Determinant, inverse, and eigenvalues

The matrix must be square for these operations.

- `np.linalg.det(A)`: determinant, a number that helps indicate whether the matrix is invertible.
- `np.linalg.inv(A)`: inverse matrix. It exists only when the determinant is nonzero (up to numerical precision).
- `np.linalg.eigvals(A)`: eigenvalues of the matrix.

The example matrix is chosen to have determinant 1, so its inverse exists.

```python
A = np.array([
    [1, 2, 3],
    [0, 1, 4],
    [5, 6, 0]
])

determinant = np.linalg.det(A)
inverse = np.linalg.inv(A)
eigenvalues = np.linalg.eigvals(A)

print("Matrix:\n", A)
print("Determinant:", determinant)
print("Inverse:\n", inverse)
print("Eigenvalues:", eigenvalues)
```

### B. Matrix multiplication: `@`, `np.matmul()`, and `np.dot()`

For matrices with shapes `(m, n)` and `(n, p)`, matrix multiplication is allowed because the inner dimensions `n` match. The output shape is `(m, p)`.

Use `A @ B` or `np.matmul(A, B)` for matrix multiplication. This differs from `A * B`, which is element-wise multiplication.

```python
A = np.array([[1, 2, 3], [4, 5, 6]])        # shape (2, 3)
B = np.array([[7, 8], [9, 10], [11, 12]])   # shape (3, 2)

product = A @ B
print("A shape:", A.shape)
print("B shape:", B.shape)
print("Product shape:", product.shape)
print(product)
```

## Assignment 7 — Reshaping and flattening

### A. Reshape to a row and a column

`reshape(1, 9)` creates a 2D row-shaped array; `reshape(9, 1)` creates a 2D column-shaped array. The total number of elements must remain 9.

```python
arr = np.arange(1, 10).reshape(3, 3)
row = arr.reshape(1, 9)
column = arr.reshape(9, 1)

print("Original:\n", arr)
print("Row shape:", row.shape, "\n", row)
print("Column shape:", column.shape, "\n", column)
```

### B. `flatten()` and reshape back

`array.flatten()` converts a multidimensional array into a **1D copy**. Changes to the flattened copy do not change the original array. Reshape it back to `(5, 5)` when it contains 25 elements.

```python
arr = np.random.randint(1, 20, size=(5, 5))
flat = arr.flatten()
restored = flat.reshape(5, 5)

print("Original:\n", arr)
print("Flattened shape:", flat.shape)
print(flat)
print("Restored:\n", restored)
```

## Assignment 8 — Fancy indexing and Boolean indexing

### A. Fancy indexing

Fancy indexing selects several positions using lists or arrays of indices. For a 2D array, `arr[row_indices, column_indices]` selects paired coordinates.

The four corners have coordinates `(0,0)`, `(0,4)`, `(4,0)`, `(4,4)`.

```python
arr = np.random.randint(1, 20, size=(5, 5))
print("Array:\n", arr)

rows = [0, 0, 4, 4]
cols = [0, 4, 0, 4]
corners = arr[rows, cols]

print("Corners (top-left, top-right, bottom-left, bottom-right):")
print(corners)
```

### B. Boolean indexing

A comparison such as `arr > 10` creates a Boolean array containing `True` where the condition is met and `False` elsewhere. Using that Boolean array as an index selects the matching elements.

`arr[arr > 10] = 10` changes every value greater than 10 to 10, directly modifying `arr`.

```python
arr = np.random.randint(1, 20, size=(4, 4))
print("Before:\n", arr)

mask = arr > 10
print("Boolean mask:\n", mask)

arr[mask] = 10
print("After capping values at 10:\n", arr)
```

## Assignment 9 — Structured arrays

A **structured array** stores records with named fields, somewhat like rows in a small table. A `dtype` describes each field's name and data type.

### A. Fields, sorting, and `np.argsort()`

- `dtype=[('name', 'U20'), ('age', 'i4'), ('weight', 'f8')]` defines a Unicode string up to 20 characters, a 32-bit integer, and a 64-bit float.
- `people['age']` accesses the age field.
- `np.argsort(values)` returns the indices that would sort the values.
- `people[np.argsort(people['age'])]` reorders records from youngest to oldest.

Common dtype codes: `U20` = Unicode string, `i4` = 4-byte integer, `f8` = 8-byte float.

```python
people_dtype = [('name', 'U20'), ('age', 'i4'), ('weight', 'f8')]
people = np.array([
    ('Amina', 22, 52.5),
    ('Rahim', 19, 67.2),
    ('Nadia', 25, 58.1),
    ('Karim', 21, 71.0)
], dtype=people_dtype)

print("People:")
print(people)
print("Ages:", people['age'])

sorted_people = people[np.argsort(people['age'])]
print("Sorted by age:")
print(sorted_people)
```

Different between `np.sort()` & `np.argsort()`:

- `np.sort()` returns the sorted values.
- `np.argsort()` returns the indices (positions) that would arrange the values in sorted order.

Let's see an example.

```python
arr = np.array([30, 10, 40, 20])

print(np.sort(arr))
print(np.argsort(arr))
```

### B. Euclidean distance between points

For a point `(x, y)`, distance from the origin is:

`sqrt(x**2 + y**2)`

- `np.sqrt()` calculates square roots.
- `** 2` squares values.
- Structured fields `points['x']` and `points['y']` provide arrays of coordinates.

For distance between two arbitrary points, use `sqrt((x2-x1)**2 + (y2-y1)**2)`.

```python
point_dtype = [('x', 'i4'), ('y', 'i4')]
points = np.array([(0, 0), (3, 4), (5, 12), (8, 15)], dtype=point_dtype)

distances_from_origin = np.sqrt(points['x']**2 + points['y']**2)

print("Points:", points)
print("Distances from origin:", distances_from_origin)
```

## Assignment 10 — Masked arrays

Masked arrays let you mark values as **excluded** from many calculations without deleting or overwriting the underlying data. Import the masked-array module as `np.ma`.

### A. Mask values greater than 10 and sum the rest

- `np.ma.masked_greater(data, 10)` masks values strictly greater than 10.
- `masked_array.mask` shows which positions are masked (`True` means excluded).
- `masked_array.sum()` sums only unmasked values.

This differs from Boolean indexing: masking keeps the array's shape and records which values are excluded.

```python
data = np.random.randint(1, 21, size=(4, 4))
masked = np.ma.masked_greater(data, 10)

print("Original data:\n", data)
print("Masked array ( -- means masked):\n", masked)
print("Mask (True means excluded):\n", masked.mask)
print("Sum of unmasked values:", masked.sum())
```

### B. Mask the diagonal and replace it with the mean of unmasked values

- `np.ma.masked_array(data, mask=mask)` creates a masked array with a custom mask.
- `np.eye(3, dtype=bool)` creates a 3×3 Boolean identity matrix: `True` on the diagonal and `False` elsewhere.
- `masked.mean()` calculates the mean of unmasked values.
- `.filled(value)` returns a regular array with masked positions replaced by `value`.

We calculate the mean before filling, so the diagonal does not affect the mean.

```python
data = np.random.randint(1, 20, size=(3, 3))
diagonal_mask = np.eye(3, dtype=bool)
masked = np.ma.masked_array(data, mask=diagonal_mask)

mean_unmasked = masked.mean()
filled = masked.filled(mean_unmasked)

print("Original:\n", data)
print("Diagonal mask:\n", diagonal_mask)
print("Masked array:\n", masked)
print("Mean of unmasked elements:", mean_unmasked)
print("After replacing masked diagonal elements:\n", filled)
```
