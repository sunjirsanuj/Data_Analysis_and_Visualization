2. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.

```python
import numpy as np
import pandas as pd
```

```python
arr = np.random.randint(1, 31, size=(6, 4))
print(arr)
```

```python
columns = list("ABCD")
df = pd.DataFrame(arr, columns=columns).astype(float)
display(df)
```

```python
df.loc[1, "A"] = np.nan
df.loc[4, "D"] = np.nan
display(df)
```

```python
droped_df = df.dropna()
display(droped_df)
```
