1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

```python
import numpy as np
import pandas as pd
```

```python
arr = np.random.randint(1, 31, size=(5, 3))
print(arr)
```

```python
columns = ["A", "B", "C"]
df = pd.DataFrame(arr, columns=columns).astype(float)
display(df)
```

```python
df.loc[1, "A"] = np.nan
df.loc[3, "B"] = np.nan
df.loc[4, "C"] = np.nan
display(df)
```

```python
means = df.mean()
print(means)
```

```python
filled = df.fillna(means)
display(filled)
```
