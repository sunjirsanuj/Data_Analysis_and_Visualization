2. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

```python
import numpy as np
import pandas as pd
```

```python
categories = [
    'Electronics',
    'Electronics',
    'Electronics',
    'Clothing',
    'Clothing',
    'Clothing'
]

subcategories = [
    'Phone',
    'Laptop',
    'Tablet',
    'Shirt',
    'Pants',
    'Shoes'
]

index = pd.MultiIndex.from_arrays(
    [categories, subcategories],
    names = ['Category', 'Subcategory']
)

df = pd.DataFrame(
    {'Sales': np.random.randint(10, 100, size=6)},
    index=index
)

display(df)
```

```python
categories_sum = df.groupby(level='Category')['Sales'].sum()
print(categories_sum)
```

```python
subcategories_sum = df.groupby(level=['Category', "Subcategory"])['Sales'].sum()
print(subcategories_sum)
```
