1. Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.

```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Marks": [85, 90, 78, 88, 92, 80],
    "Age": [20, 21, 19, 22, 20, 21]
}

index = [
    ['CSE', 'CSE', 'CSE', 'EEE', 'EEE', 'EEE'],
    ['Rahim', 'Karim', 'Sakib', 'Nadia', 'Mim', 'Riya']
]

df = pd.DataFrame(
    data,
    index=index
)

df.index.names = ["Department", "Student"]
display(df)
```

```python
# Operation 1: Access a row using both index levels
print(df.loc[("CSE", "Rahim")])
```

```python
# Operation 2: Access all rows from one department
print(df.loc["CSE"])
```

```python
# Operation 3: Access a specific column for one department
print(df.loc["CSE", "Marks"])
```

```python
# Operation 4: Access a specific value
print(df.loc[("EEE", 'Mim'), 'Age'])
```

```python
# Operation 5: Slice rows using the first index level
df = df.sort_index()
print(df.loc["CSE":"EEE"])
```

```python
# Operation 6: Select rows using IndexSlice
idx = pd.IndexSlice
print(df.loc[idx['CSE', 'Karim':'Sakib'], :])
```

```python
# Operation 7: Access a row using .iloc[]
print(df.iloc[0])
```
