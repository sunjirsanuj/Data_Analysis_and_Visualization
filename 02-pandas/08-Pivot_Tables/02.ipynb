2. Create a Pandas DataFrame with columns 'Year', 'Quarter', and 'Revenue'. Create a pivot table to compute the mean 'Revenue' for each 'Quarter' by 'Year'.

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Year': [
        2024, 2024, 2024, 2024,
        2025, 2025, 2025, 2025
    ],
    'Quarter': [
        'Q1', 'Q2', 'Q3', 'Q4',
        'Q1', 'Q2', 'Q3', 'Q4'
    ],
    'Revenue': [
        10000, 15000, 12000, 18000,
        14000, 16000, 20000, 22000
    ]
}

df = pd.DataFrame(data)
display(df)
```

```python
pivot = pd.pivot_table(
    df,
    values='Revenue',
    index='Quarter',
    columns='Year',
    aggfunc='mean'
)
display(pivot)
```
