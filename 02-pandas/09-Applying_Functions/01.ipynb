1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.

```python
import numpy as np
import pandas as pd
```

```python
arr = np.random.randint(1, 100, size=(5, 3))
df = pd.DataFrame(
    arr,
    columns=['A', 'B', 'C']
)
display(df)
```

```python
def double(x):
    return x*2

result = df.map(double)
display(result)
```
