2. Create two Pandas DataFrames with different columns. Concatenate the DataFrames along the rows and along the columns.

```python
import numpy as np
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "Name": ("Sunjir", "Azim", "Karim", "Arafat"),
    "Age": (21, 21, 23, 20)
})

df2 = pd.DataFrame({
    "City": ("Sylhet", "Korea", "Dhaka", "London"),
    "Mark": (90, 95, 80, 100)
})

display(df1)
display(df2)
```

```python
row_wise = pd.concat([df1, df2], axis=0)
column_wise = pd.concat([df1, df2], axis=1)
display(row_wise)
display(column_wise)
```
