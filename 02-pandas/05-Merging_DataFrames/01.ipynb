1. Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

```python
import numpy as np
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "ID": (1, 2, 3, 4),
    "Name": ("Sunjir", "Azim", "Karim", "Arafat")
})

df2 = pd.DataFrame({
    "ID": (1, 2, 3, 4),
    "Mark": (80, 94, 100, 50)
})

display(df1)
display(df2)
```

```python
df_merged = pd.merge(df1, df2, on="ID")
display(df_merged)
```
