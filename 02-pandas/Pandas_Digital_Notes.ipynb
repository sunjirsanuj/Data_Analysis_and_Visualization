## Quick reference

- `date_range()` creates dates; `resample()` groups time intervals; `rolling()` calculates rolling windows.

# Assignment 3 — Data Cleaning

## 3.1 Fill NaN with each column's mean
`mean()` skips missing values by default. `fillna(df.mean())` fills each column using its own mean.


|Code|What it calculates|
|---|---|
|`df.mean()`|Column-wise mean|
|`df.mean(axis=0)`|Column-wise mean|
|`df.mean(axis=1)`|Row-wise mean|


````python
df = pd.DataFrame(np.random.randint(1, 20, (5, 3)).astype(float), columns=["A", "B", "C"])
df.loc[1, "A"] = np.nan
df.loc[3, "B"] = np.nan
df.loc[4, "C"] = np.nan
print("With NaN:")
display(df)
print("Column means:")
print(df.mean())
print("Filled:")
display(df.fillna(df.mean()))
````

# 3.2 Why does it ignore NaN values?

by default `df.mean()` uses `skipna=True`. That means missing values are excluded from both the sum and the count.

## 3.3 Drop rows with any NaN
`dropna()` removes rows with at least one missing value by default. `axis=1` would drop columns instead.


````python
df = pd.DataFrame(np.random.randint(1, 20, (6, 4)).astype(float), columns=list("ABCD"))
df.loc[1, "B"] = np.nan
df.loc[4, "D"] = np.nan
display(df)
display(df.dropna())
````

## `choice()` -

In Python, `choice()` is commonly used to randomly select one item from a collection.
There are two versions worth knowing:

- `random.choice()` from Python's built-in `random` module.
- `np.random.choice()` from NumPy.

### Syntax:

````python
np.random.choice(a, size=None, replace=True, p=None)
````

|Parameter|Meaning|
|---|---|
|`a`|The list, array, or integer population to select from|
|`size`|Number of items to select|
|`replace`|Whether an item can be selected more than once|
|`p`|Optional probability of selecting each item|

Example 1: Select one random item

````python
data = [10, 20, 30, 40, 50]

result = np.random.choice(data)

print(result)
````

Example 2: Select multiple random items

````python
data = [10, 20, 30, 40, 50]

result = np.random.choice(data, size=3)

print(result)
````

Example 3: Select without repetition

````python
data = [10, 20, 30, 40, 50]

result = np.random.choice(data, size=3, replace=False)

print(result)
````

## Pandas `agg()` -

In Pandas, agg() stands for aggregation. It allows you to apply one or more functions to a Series or DataFrame and get summarized results.

For example, suppose you have students' marks and want to calculate the average, highest mark, lowest mark, and total marks.
Instead of writing separate calculations, you can use agg() to perform them together.

Understanding `agg()` with a simple example:

````python
import pandas as pd
df = pd.DataFrame({
    'Math': [80, 90, 70, 85],
    'English': [75, 85, 95, 80],
    'Science': [90, 80, 85, 95]
})

display(df)

print(df.agg(['mean', 'max', 'min']))
print(df.agg('mean'))
````

### Applying different functions to different columns

````python
result = df.agg({
    'Math': ['mean', 'max'],
    'English': ['min', 'max'],
    'Science': ['sum', 'mean']
})

print(result)
````

# Assignment 6 — Time Series Analysis

## 6.1 Monthly mean using a datetime index
`date_range()` creates dates; `resample("MS")` groups by month-start; `mean()` calculates the monthly mean.


````python
dates = pd.date_range("2024-01-01", periods=90, freq="D")
df = pd.DataFrame({"Value": np.random.randint(1, 100, len(dates))}, index=dates)
display(df.head())
display(df.resample("MS").mean())
````

### `pd.data_range()` in Pandas

`pd.data_range()` is a Pandas function used to generate a sequence of dates or timestamps at a fixed frequency. It is commonly used to create a datetime index for time-series data.

Syntax:

````python
pd.date_range(start=None, end=None, periods=None, freq='D')
````

Parameters

- `start`: The starting date of the sequence.
- `end`: The ending date of the sequence.
- `periods`: The number of dates or timestamps to generate.
- `freq`: The frequency or time interval between consecutive dates.

## 6.2 Seven-day rolling mean
With one row per day, `rolling(window=7)` uses seven consecutive daily observations. `min_periods=7` keeps the first six means as NaN until a full window exists.

The `rolling()` method creates a sliding window over the data.
- `rolling()` moves a window across the observations.
- `window=7` specifies that each window contains 7 observations.
- As the window moves forward, the oldest observation is removed and the next observation is included.


````python
dates = pd.date_range("2021-01-01", "2021-12-31", freq="D")
df = pd.DataFrame({"Value": np.random.randint(1, 100, len(dates))}, index=dates)
df["RollingMean7Days"] = df["Value"].rolling(window=7, min_periods=7).mean()
display(df.head(10))
````

### What is `rolling()`?

In Pandas, the rolling() method is used to create a sliding window over a sequence of data. It allows us to perform calculations on a fixed number of consecutive observations, such as calculating the mean, sum, minimum, maximum, or standard deviation.

# Assignment 7 — MultiIndex DataFrame

## 7.1 Create a hierarchical index and perform indexing/slicing
A MultiIndex has more than one level of labels.


````python
idx = pd.MultiIndex.from_product([["Group1", "Group2"], ["A", "B", "C"]],
                                   names=["Group", "Item"])
df = pd.DataFrame({"Score": np.random.randint(50, 100, 6)}, index=idx)
display(df)
display(df.loc["Group1"])
display(df.loc[("Group2", "B")])
display(df.loc[("Group1", "A"):("Group1", "B")])
````

## 7.2 Sum values by Category and SubCategory
`groupby(level=...)` groups by named MultiIndex levels.


````python
idx = pd.MultiIndex.from_product([["Food", "Drink"], ["Local", "Imported"]],
                                   names=["Category", "SubCategory"])
df = pd.DataFrame({"Value": np.random.randint(10, 100, 4)}, index=idx)
display(df)
display(df.groupby(level="Category")["Value"].sum())
display(df.groupby(level=["Category", "SubCategory"])["Value"].sum())
````

# Assignment 8 — Pivot Tables

## 8.1 Sum Value for each Category by Date
`index` becomes the rows, `columns` become separate columns, `values` selects the data, and `aggfunc` specifies the calculation.


````python
df = pd.DataFrame({
    "Date": ["2024-01-01", "2024-01-01", "2024-01-02", "2024-01-02", "2024-01-03", "2024-01-03"],
    "Category": ["A", "B", "A", "B", "A", "B"],
    "Value": [10, 20, 15, 25, 12, 18]
})
display(df)
display(pd.pivot_table(df, index="Date", columns="Category", values="Value", aggfunc="sum"))
````

# Assignment 9 — Applying Functions

## 9.1 Double all DataFrame values
For numeric data, `df * 2` is the simplest vectorized solution. `apply()` can also apply an operation to each column.


````python
df = pd.DataFrame(np.random.randint(1, 20, (5, 3)), columns=["A", "B", "C"])
display(df)
display(df * 2)
display(df.apply(lambda column: column * 2))
````

## 9.2 Lambda function to sum existing columns into a new column
`axis=1` means apply the function to each row. For simple totals, `sum(axis=1)` is an easier alternative.


````python
df = pd.DataFrame(np.random.randint(1, 20, (6, 3)), columns=["A", "B", "C"])
df["Total"] = df.apply(lambda row: row["A"] + row["B"] + row["C"], axis=1)
display(df)
# Alternative: df["Total"] = df[["A", "B", "C"]].sum(axis=1)
````

# Assignment 10 — Working with Text Data

## 10.1 Convert strings to uppercase
`.str.upper()` applies uppercase conversion to each string in the Series.


````python
words = pd.Series(["pandas", "numpy", "data science", "python", "hello world"])
display(words.to_frame("Original"))
display(words.str.upper().to_frame("Uppercase"))
````
