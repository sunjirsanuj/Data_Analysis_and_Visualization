1. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Set the index to be the first columcn.

```python
import numpy as np 
import pandas as pd 
```

```python
arr = np.random.randint(1, 100, size=(6, 4))
df = pd.DataFrame(arr, columns=["A", "B", "C", "D"])
print(df)
df = df.set_index("A")
print(df)
```
